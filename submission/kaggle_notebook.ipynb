# Optiver - Trading at the Close: Submission Notebook

Loads the trained LightGBM model from the attached dataset and runs inference via the `iter_test` API.

```python
import shutil, os

# 1. Copy competition data (optiver2023 + test.csv)
SRC = "/kaggle/input/competitions/optiver-trading-at-the-close/"
DST = "/kaggle/working/"
for item in os.listdir(SRC):
    s = os.path.join(SRC, item); d = os.path.join(DST, item)
    if os.path.isdir(s) and not os.path.exists(d):
        shutil.copytree(s, d)
    elif os.path.isfile(s) and not os.path.exists(d):
        shutil.copy2(s, d)

# 2. Copy model (txt format) + inference script
DS = "/kaggle/input/datasets/eliansun127/optiver-trading-at-the-close-model/"
for f in ["lgb_model.txt", "notebook_submission.py"]:
    shutil.copy(DS + f, DST + f)

print("Setup done.")
print("Files:", sorted(os.listdir(DST)))
```
